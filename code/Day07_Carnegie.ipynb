```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score

print(pd.__version__)
print(np.__version__)

url = ("https://raw.githubusercontent.com/ryanaxiom/Applied-ML/main/data/carnegie_data.csv")

carnegie_raw = pd.read_csv(url, encoding="cp1252")
carnegie = carnegie_raw.copy()
carnegie["research"] = carnegie["serd"] + carnegie["nonserd"]
d = carnegie[carnegie["research"].notna()].copy()
d["med_yn"] = (d["medical"] == 1 ).astype(int)
y = d["research"]

model2 = LinearRegression()
model2.fit(d[["fallenr20","med_yn"]],y)
```

```python
r2_score(y,model2.predict(d[["fallenr20","med_yn"]]))
```

```python
newx = pd.DataFrame({"fallenr20": [11254, 32277], "med_yn": [0,1]})
model2.predict(newx)
```

```python
model3 = LinearRegression()
model3.fit(d[["fallenr20","med_yn","facnum"]],y)
model3.coef_, model3.intercept_
```

```python
r2_score(y,model3.predict(d[["fallenr20","med_yn","facnum"]]))
```

```python
model3 = LinearRegression()
model3.fit(d[["fallenr20","med_yn","stem_rsd"]],y)
```

```python
print(r2_score(y,model3.predict(d[["fallenr20","med_yn","stem_rsd"]])))
model3.coef_, model3.intercept_
```

```python
pile = ["fallenr20","totdeg","facnum","stem_rsd","med_yn"]
```

```python
model5 = LinearRegression()
model5.fit(d[pile],y)
```

```python
r2_score(y,model5.predict(d[pile]))
```

```python
m5pred = model5.predict[d[pile]]
np.sqrt(np.mean((y-m5pred)**2))
```

```python
model5.coef_,model5.intercept_
```

```python
r2_score(d[["fallenr20"]],d[["totdeg"]])
```

```python
rng = np.random.default_rng(309)
idx = rng.permutation(len(d))
A, B = d.iloc[idx[:151]],d.iloc[idx[151:]]
mA = LinearRegression()
mA.fit(A[pile],A["research"])
mB = LinearRegression()
mB.fit(B[pile],B["research"])
```

```python
print("model A: ",mA.coef_,mA.intercept_)
print("model B: ",mB.coef_,mB.intercept_)
```

```python
pile2 = ["fallenr20","facnum","stem_rsd","med_yn"]
rng = np.random.default_rng(309)
idx = rng.permutation(len(d))
A, B = d.iloc[idx[:151]],d.iloc[idx[151:]]
mA = LinearRegression()
mA.fit(A[pile2],A["research"])
mB = LinearRegression()
mB.fit(B[pile2],B["research"])
print("model A: ",mA.coef_,mA.intercept_)
print("model B: ",mB.coef_,mB.intercept_)
```

```python
np.sqrt(r2_score(mA.predict(d[pile]),mB.predict(d[pile])))
```
