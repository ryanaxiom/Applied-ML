```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

print(pd.__version__)
print(np.__version__)

url = ("https://raw.githubusercontent.com/ryanaxiom/Applied-ML/main/data/carnegie_data.csv")

carnegie_raw = pd.read_csv(url, encoding="cp1252")
carnegie = carnegie_raw.copy()
carnegie["research"] = carnegie["serd"] + carnegie["nonserd"]
d = carnegie[carnegie["research"].notna()].copy()
d.shape
```

```python
(1-4.33)**2 + (2-4.33)**2 + (10-4.33)**2
```

```python
y=d["research"]
grid = np.linspace(y.min(),y.max(),4000)
sq = [np.sum((y-c)**2) for c in grid]
ab = [np.sum(np.abs(y-c)) for c in grid]
grid[np.argmin(sq)], grid[np.argmin(ab)]
```

```python
y.mean(), y.median()
```

```python
x = d["fallenr20"]

r = x.corr(y)
b1 = r * y.std() / x.std()
b0 = y.mean() - b1 * x.mean()
r, b1, b0
```

```python
from sklearn.linear_model import LinearRegression
model = LinearRegression()
model.fit(d[["fallenr20"]],d[["research"]])
model.coef_, model.intercept_
```
