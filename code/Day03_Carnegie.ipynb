```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

url = ("https://raw.githubusercontent.com/ryanaxiom/Applied-ML/main/data/carnegie_data.csv")

carnegie = pd.read_csv(url, encoding="cp1252")
carnegie["research"] = carnegie["serd"] + carnegie["nonserd"]
y=carnegie["research"].dropna()
```

```python
print(y.mean())
print(y.median())
print(y.std())
#y.describe()
```

```python
np.sqrt(np.mean((y-y.mean())**2))
```

```python
carnegie["research"].hist()
plt.xlabel("Research Spending ($ thousands)")
plt.show()
```

```python
carnegie["research"].hist(bins=40)
plt.xlabel("Research Spending ($ thousands)")
plt.xscale("log")
plt.show()
```

```python
bins = np.logspace(np.log10(y.min()), np.log10(y.max()), 30)
carnegie["research"].hist(bins=bins)
plt.xlabel("Research Spending ($ thousands)")
plt.xscale("log")
plt.show()
```

```python
d = carnegie[carnegie["research"].notna()]
d.columns.tolist()
```

```python
plt.scatter(d["fallenr20"],d["research"], s=14, alpha=.6)
plt.xlabel("Enrollment")
plt.ylabel("Research Spending ($ thousands)")
plt.show()
```

```python
plt.scatter(d["fallenr20"],d["research"],s=14,alpha=.6)
plt.yscale("log")
plt.show()
```

```python
has_med = d[d["medical"] == 1]
no_med = d[d["medical"] == 2]

plt.scatter(no_med["fallenr20"], no_med["research"],label="no medical school")
plt.scatter(has_med["fallenr20"], has_med["research"],label="medical school")
plt.yscale("log")
plt.legend()
plt.show()
```

```python
import plotly.express as px

px.scatter_3d(d, x="fallenr20", y="totdeg", z="research", color="medical", log_z=True)
```
